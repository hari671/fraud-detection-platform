```python
import pandas as pd              # for loading and working with tables/dataframes
import numpy as np               # for numerical operations
import matplotlib.pyplot as plt  # for basic plots/charts

pd.set_option("display.max_columns", 500)  # lets pandas show up to 500 columns when printing
pd.set_option("display.max_rows", 200)     # lets pandas show more rows before truncating
```

```python
# Load the main transaction dataset
train_trans = pd.read_csv("../data/raw/train_transaction.csv")

# Load the identity dataset, which has extra user/device-related info
train_id = pd.read_csv("../data/raw/train_identity.csv")

# Print the size of each dataset so we know what we loaded
print("train_transaction shape:", train_trans.shape)
print("train_identity shape:", train_id.shape)
```

```python
# Merge the transaction and identity tables using TransactionID
# how="left" means: keep every transaction, and add identity info if it exists
df = train_trans.merge(train_id, on="TransactionID", how="left")

# Print the shape of the merged dataset
print("merged shape:", df.shape)

# Show the first 5 rows so we can inspect what the merged data looks like
df.head()
```

```python
# Show a summary of the dataframe:
# column names, non-null counts, and data types
df.info()
```

```python
# Count how many rows belong to each class: 0 = not fraud, 1 = fraud
fraud_counts = df["isFraud"].value_counts()

# Compute the overall fraud rate as a decimal
fraud_rate = df["isFraud"].mean()

# Print both raw counts and percentage
print("Fraud counts:")
print(fraud_counts)

print("\nFraud rate:", fraud_rate)
print("Fraud rate (%):", fraud_rate * 100)
```

```python
# Make a bar chart showing how many fraud vs non-fraud transactions there are
fraud_counts.plot(kind="bar")

# Add chart labels/title so it is easy to read
plt.title("Fraud vs Non-Fraud Transactions")
plt.xlabel("isFraud")
plt.ylabel("Count")
plt.xticks(rotation=0)

# Display the chart
plt.show()
```

```python
# Calculate what percent of each column is missing
missing_pct = df.isnull().mean().sort_values(ascending=False) * 100

# Convert the result into a clean dataframe for easier reading
missing_df = missing_pct.reset_index()
missing_df.columns = ["column", "missing_percent"]

# Show the top 30 columns with the most missing values
missing_df.head(30)
```

```python
# Get a list of all numeric columns
numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()

# Get a list of all categorical/text columns
categorical_cols = df.select_dtypes(include=["object"]).columns.tolist()

# Print how many of each type we have
print("Number of numeric columns:", len(numeric_cols))
print("Number of categorical columns:", len(categorical_cols))

# Print the categorical column names
print("\nCategorical columns:")
print(categorical_cols)
```

```python
# These are some key columns commonly used in fraud analysis
key_cols = [
    "isFraud", "TransactionAmt", "ProductCD",
    "card1", "card2", "card3", "card4", "card5", "card6",
    "addr1", "addr2", "P_emaildomain", "R_emaildomain"
]

# Keep only the columns that actually exist in the dataframe
existing_key_cols = [col for col in key_cols if col in df.columns]

# Display the first few rows of those key columns
df[existing_key_cols].head(10)
```

```python
# Show descriptive statistics for numeric columns
# This includes count, mean, std, min, max, and quartiles
df.describe()
```

```python
# Show descriptive stats specifically for transaction amount
print(df["TransactionAmt"].describe())

# Plot a histogram of transaction amounts
df["TransactionAmt"].hist(bins=50)

# Add labels/title
plt.title("Transaction Amount Distribution")
plt.xlabel("TransactionAmt")
plt.ylabel("Frequency")

# Show the plot
plt.show()
```

```python
# Compute summary statistics for transaction amount grouped by fraud label
df.groupby("isFraud")["TransactionAmt"].describe()
```

```python
# Count how often each ProductCD category appears
print(df["ProductCD"].value_counts(dropna=False))

# Plot category frequencies
df["ProductCD"].value_counts(dropna=False).plot(kind="bar")

# Add labels/title
plt.title("ProductCD Value Counts")
plt.xlabel("ProductCD")
plt.ylabel("Count")
plt.xticks(rotation=0)

# Show the plot
plt.show()
```

```python
# Compute the average fraud rate for each ProductCD category
# Since isFraud is 0/1, the mean gives the fraud percentage as a decimal
fraud_by_product = df.groupby("ProductCD")["isFraud"].mean().sort_values(ascending=False)

# Print the fraud rate by product category
print(fraud_by_product)

# Plot the fraud rate by ProductCD
fraud_by_product.plot(kind="bar")

# Add labels/title
plt.title("Fraud Rate by ProductCD")
plt.xlabel("ProductCD")
plt.ylabel("Fraud Rate")
plt.xticks(rotation=0)

# Show the plot
plt.show()
```

```python
# Show the most common purchaser email domains
print("Top P_emaildomain values:")
print(df["P_emaildomain"].value_counts(dropna=False).head(15))

print("\nTop R_emaildomain values:")
print(df["R_emaildomain"].value_counts(dropna=False).head(15))
```

```python
# Find the top 10 most common purchaser email domains
top_p_domains = df["P_emaildomain"].value_counts().head(10).index

# Filter the dataframe to only those top domains
temp = df[df["P_emaildomain"].isin(top_p_domains)]

# Compute fraud rate for each of those domains
fraud_by_domain = temp.groupby("P_emaildomain")["isFraud"].mean().sort_values(ascending=False)

# Print the fraud rate by domain
print(fraud_by_domain)

# Plot the fraud rates
fraud_by_domain.plot(kind="bar")

# Add labels/title
plt.title("Fraud Rate by Top Purchaser Email Domains")
plt.xlabel("P_emaildomain")
plt.ylabel("Fraud Rate")
plt.xticks(rotation=45)

# Show the plot
plt.show()
```
