```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Put all model results side by side
comparison_df = pd.DataFrame({
    "Model": ["Logistic Regression", "XGBoost", "CatBoost"],
    "Fraud Precision": [0.07, 0.31, 0.24],
    "Fraud Recall": [0.67, 0.83, 0.82],
    "Fraud F1": [0.12, 0.45, 0.37],
    "ROC AUC": [0.7443, 0.9513, 0.9365],
    "PR AUC": [0.1325, 0.7136, 0.6436],
    "Predicted Frauds": [40976, 11157, 14281]
})

comparison_df
```

```python
# Make model names the row labels
comparison_df = comparison_df.set_index("Model")

# Show the cleaned comparison table
comparison_df
```

```python
# Compare the three most important model-quality metrics
comparison_df[["Fraud F1", "ROC AUC", "PR AUC"]].plot(kind="bar")

# Add chart labels/title
plt.title("Model Performance Comparison")
plt.ylabel("Score")
plt.xticks(rotation=0)

# Show the plot
plt.show()
```

```python
# Compare fraud precision and recall across models
comparison_df[["Fraud Precision", "Fraud Recall"]].plot(kind="bar")

# Add chart labels/title
plt.title("Fraud Precision vs Recall by Model")
plt.ylabel("Score")
plt.xticks(rotation=0)

# Show the plot
plt.show()
```

```python
# Compare how many frauds each model predicted
comparison_df["Predicted Frauds"].plot(kind="bar")

# Add chart labels/title
plt.title("Predicted Fraud Counts by Model")
plt.ylabel("Count")
plt.xticks(rotation=0)

# Show the plot
plt.show()
```

```python
# Find the model with the highest PR AUC
best_model = comparison_df["PR AUC"].idxmax()

print("Best model based on PR AUC:", best_model)
```

## Model Comparison Summary

- Logistic Regression served as the baseline model but produced a very large number of false positives and weak PR AUC.
- CatBoost improved substantially over the baseline and performed well on the mixed-type transaction data.
- XGBoost achieved the strongest overall performance across fraud precision, fraud F1, ROC AUC, PR AUC, and false-positive tradeoff.
- Based on model comparison results, XGBoost was selected as the final fraud detection model for threshold tuning and deployment.
